```python
import numpy as np
a = np.array([1, 0])
b = np.array([0, 1])
print(np.kron(a, b))   # [0 1 0 0]
print(np.kron(b, a))   # [0 0 1 0]

H = (1/np.sqrt(2)) * np.array([[1, 1],
                               [1, -1]])
z = H@H
print(z)

np.allclose(H @ H, np.eye(2))   # True

y = H@a
print(y)

np.abs(y)**2


print(H@b)
np.abs(H@b)**2

print(H @ (H @ a))   # [1, 0]: back to |0⟩
print(H @ (H @ b))   # [0, 1]: back to |1⟩

S = np.array([[1, 0], [0, 1j]])
S.conj().T

np.allclose(S.conj().T @ S, np.eye(2))   # True

def is_unitary(U):
    n = U.shape[0]
    return np.allclose(U.conj().T @ U, np.eye(n))

X = np.array([[0, 1], [1, 0]])
vals, vecs = np.linalg.eig(X)
print(vals)
print(vecs)

np.allclose(X @ vecs[:, 0], vals[0] * vecs[:, 0])
np.allclose(X @ vecs[:, 1], vals[1] * vecs[:, 1])

np.kron(H, np.eye(2))
```
